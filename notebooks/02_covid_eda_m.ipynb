Cargar datos mundiales

```python
import pandas as pd

# Listado limpio sin espacios invisibles
columnas_seleccionadas = [
    'iso_code',
    'continent',
    'location',  # País
    'date',  # Fecha
    'total_cases',
    'new_cases',
    'total_deaths',  # Muertes
    'new_deaths',
    'total_vaccinations',  # Vacunas administradas
    'people_vaccinated',  # Personas con al menos 1 dosis
    'people_fully_vaccinated',  # Personas con pauta completa
    'total_boosters',  # Dosis de refuerzo
    'population',  # Población del país
    'gdp_per_capita',  # Riqueza del país
    'life_expectancy',  # Esperanza de vida
    'stringency_index',  # Rigidez de restricciones del gobierno
]

url_owid = 'https://raw.githubusercontent.com/owid/covid-19-data/master/public/data/owid-covid-data.csv'

# Cargar el archivo filtrando las 16 columnas
df_mundial = pd.read_csv(url_owid, usecols=columnas_seleccionadas, parse_dates=['date'])

# Mostrar resultados
print(f'Dimensiones del DataFrame: {df_mundial.shape}')
display(df_mundial.head())
display(df_mundial.tail(10))
```
